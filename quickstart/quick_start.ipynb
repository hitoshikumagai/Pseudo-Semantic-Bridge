```python
import sys

if sys.version_info[0] < 3:
    raise RuntimeError("Please run this notebook with a Python 3 kernel.")

from pathlib import Path
import os

# Notebookの起動場所に依存せず、repoルートを解決する
search_roots = [Path.cwd()]
search_roots.extend(list(Path.cwd().parents))

PROJECT_ROOT = None
for candidate in search_roots:
    if (candidate / "src").exists() and (candidate / "configs").exists():
        PROJECT_ROOT = candidate
        break

if PROJECT_ROOT is None:
    raise RuntimeError("Project root not found. Please open notebook inside the repository.")

os.chdir(str(PROJECT_ROOT))
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from src.bridge.builder import build_all_configs
from src.adapter.outlook import OutlookAdapter
from src.engine.core import GenericEtlEngine
from src.web.app_logic import run_pipeline_baseline

MAIN_JSON_PATH = Path("configs/accounting/invoice_bot_v2.json")

print("[Bridge] compile + [Engine] run (notebook baseline)")
pipeline_summary = run_pipeline_baseline(
    build_fn=build_all_configs,
    config_path=MAIN_JSON_PATH,
    adapter_factory=OutlookAdapter,
    engine_factory=GenericEtlEngine,
)
print("status:", pipeline_summary.get("status"))
if pipeline_summary.get("error"):
    print("error:", pipeline_summary.get("error"))
pipeline_summary
```
